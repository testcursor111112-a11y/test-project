```python
import os
from dotenv import load_dotenv
from langchain_google_genai import ChatGoogleGenerativeAI
load_dotenv()


os.environ['GOOGLE_API_KEY'] = os.getenv('GOOGLE_API_KEY') or os.getenv('GOOGLE_API_KEY_1')

model = ChatGoogleGenerativeAI(
    model='gemini-3.1-flash-lite'
)
```

```python
from pydantic import BaseModel, Field

class Movie(BaseModel):
    title:str=Field(description='the title of the movie')
    year:int=Field(description="the released date of the movie")
    director:str=Field(description='the director of the movie')
```

```python
model_with_structure = model.with_structured_output(Movie, include_raw=True)

res = model_with_structure.invoke("give me a gujrati movie released in 2001")

res
```

```python
from typing_extensions import TypedDict
from langchain.agents import create_agent
from langchain_google_genai import ChatGoogleGenerativeAI

class ContactInfo(TypedDict):
    """Contact information of the person"""
    name: str # the name of the person
    email: str # the email of the person
    phone: str # the phone of the person


llm = ChatGoogleGenerativeAI(
    model="gemini-3.1-flash-lite"
)


agent = create_agent(
    model=llm,
    response_format=ContactInfo
)

result = agent.invoke({
    "messages": [{"role":"user", "content": "Extract contact info fomr: John Doe, john@example.com, 9427514709"}]
})

result['structured_response']

# model_with_structure = model.with_structured_output(ContactInfo, include_raw=True)

# res = model.invoke("Extract contact info fomr: John Doe, john@example.com, 9427514709")
# res.content 
```
