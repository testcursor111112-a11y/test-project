```python
# how to creat a tool and bind it with chat model

from langchain.tools import tool

@tool
def get_weather(location:str)-> str:
    """Get the weather at a location"""
    return f"It's sunny in {location}"

model_with_init = model_with_init.bind_tools([get_weather])
```

```python
resp =  model_with_init.invoke('what is the weather in india?')
resp 
```
