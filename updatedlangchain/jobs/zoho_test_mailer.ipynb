# Zoho SMTP Test Mailer

Sends a single test email via Zoho SMTP using an App Password (STARTTLS on port 587).

```python
import os
from pathlib import Path
from dotenv import load_dotenv
import smtplib
from email.mime.multipart import MIMEMultipart
from email.mime.text import MIMEText

# Load repo-root .env (same file the other notebooks use)
load_dotenv(Path.cwd().parents[1] / '.env')

# Configuration from environment
SMTP_SERVER = os.getenv("ZOHO_SMTP_SERVER", "smtp.zoho.in")
SMTP_PORT = int(os.getenv("ZOHO_SMTP_PORT", "587"))  # 587 for TLS
SENDER_EMAIL = os.getenv("ZOHO_SENDER_EMAIL")
APP_PASSWORD = os.getenv("ZOHO_APP_PASSWORD")
RECEIVER_EMAIL = os.getenv("ZOHO_RECEIVER_EMAIL")

assert SENDER_EMAIL and APP_PASSWORD and RECEIVER_EMAIL, "Missing ZOHO_* keys in .env"
print("Zoho configured:", SENDER_EMAIL, "->", RECEIVER_EMAIL)
```

```python
# 1. Create the email headers and content
message = MIMEMultipart()
message["From"] = SENDER_EMAIL
message["To"] = RECEIVER_EMAIL
message["Subject"] = "Test Email from Zoho SMTP"

body = "Hello! This email was sent securely from Python using Zoho SMTP and an App Password."
message.attach(MIMEText(body, "plain"))
```

```python
try:
    # 2. Establish a secure connection with the server
    print("Connecting to server...")
    server = smtplib.SMTP(SMTP_SERVER, SMTP_PORT)
    server.starttls()  # Upgrade the connection to secure TLS

    # 3. Login and send email
    print("Logging in...")
    server.login(SENDER_EMAIL, APP_PASSWORD)

    print("Sending email...")
    server.sendmail(SENDER_EMAIL, RECEIVER_EMAIL, message.as_string())
    print("\u2705 Email sent successfully!")

except Exception as e:
    print(f"\u274c Error occurred: {e}")

finally:
    # 4. Clean up and close connection
    if 'server' in locals():
        server.quit()
        print("Connection closed.")
```
