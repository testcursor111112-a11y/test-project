# Job Application Mail Generator

Generates a customized job application email for any company:
1. Loads my resume (PDF in this folder)
2. Researches the company on the internet (DuckDuckGo search tool)
3. Writes a tailored email matching my skills to the company

```python
import os
from dotenv import load_dotenv
from langchain_google_genai import ChatGoogleGenerativeAI

load_dotenv()
os.environ['GOOGLE_API_KEY'] = os.getenv('GOOGLE_API_KEY')

model = ChatGoogleGenerativeAI(
    model='gemini-3.1-flash-lite'
)
```

## 1. Load resume from PDF

```python
from langchain_community.document_loaders import PyPDFLoader

loader = PyPDFLoader('resume.pdf')
pages = loader.load()
resume_text = '\n'.join(page.page_content for page in pages)

print(resume_text[:500])
```

## 2. Internet search tool for company research

```python
from langchain_community.tools import DuckDuckGoSearchRun

search = DuckDuckGoSearchRun()

# quick test
print(search.invoke('Lanatus systems')[:3000])
```

## 3. Agent: research company + write customized mail

```python
from langchain.agents import create_agent

system_prompt = f"""You are a job application assistant for the candidate whose resume is below.

When the user gives a company name (and optionally a role):
1. Use the search tool to research the company: what it does, its products,
   tech stack, recent news, mission and culture. Search multiple times if needed.
2. Write a customized job application email:
   - Subject line
   - Greeting to the hiring team
   - 1st paragraph: why this company specifically (use real researched facts)
   - 2nd paragraph: match the candidate's strongest relevant skills, projects
     and experience from the resume to what the company does
   - Closing with a call to action and the candidate's contact details from the resume
Keep it concise (under 250 words), professional, and specific — no generic filler.

CANDIDATE RESUME:
{resume_text}
"""`

agent = create_agent(
    model=model,
    tools=[search],
    system_prompt=system_prompt
)
```

```python
def generate_application_mail(company: str, role: str = 'Software Engineer') -> str:
    result = agent.invoke({
        'messages': [{
            'role': 'user',
            'content': f'Write a job application email to {company} for the role of {role}.'
        }]
    })
    return result['messages'][-1].content
```

## 4. Try it

```python
mail = generate_application_mail('Dev Information Technology Limited', 'AI Engineer')
print(mail[0]['text'])
```
